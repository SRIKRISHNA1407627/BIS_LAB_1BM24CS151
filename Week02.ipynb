```python
import random
import copy
import math

# -------------------------------------------------------------
# 1. SETUP THE DELIVERY ENVIRONMENT (Coordinates & Distances)
# -------------------------------------------------------------
customer_coords = {
    0: (0, 0),    # Depot
    1: (2, 7),    # Customer 1
    2: (8, 7),    # Customer 2
    3: (1, 1),    # Customer 3
    4: (7, 9),    # Customer 4
    5: (10, 8)     # Customer 5
}

def calculate_distance(coord1, coord2):
    return math.sqrt((coord1[0] - coord2[0])**2 + (coord1[1] - coord2[1])**2)

def evaluate_fitness(route):
    total_dist = 0
    total_dist += calculate_distance(customer_coords[0], customer_coords[route[0]])
    for i in range(len(route) - 1):
        total_dist += calculate_distance(customer_coords[route[i]], customer_coords[route[i+1]])
    total_dist += calculate_distance(customer_coords[route[-1]], customer_coords[0])
    return total_dist

# -------------------------------------------------------------
# 2. DISCRETE OPERATORS (The core mechanics of Discrete PSO)
# -------------------------------------------------------------
def get_swap_sequence(current_route, target_route):
    temp_route = copy.copy(current_route)
    swap_sequence = []
    for i in range(len(temp_route)):
        if temp_route[i] != target_route[i]:
            target_idx = temp_route.index(target_route[i])
            swap_sequence.append((i, target_idx))
            temp_route[i], temp_route[target_idx] = temp_route[target_idx], temp_route[i]
    return swap_sequence

def apply_swaps(route, swap_sequence):
    """Applies all valid swap operators in a sequence to a route."""
    new_route = copy.copy(route)
    for swap in swap_sequence:
        idx1, idx2 = swap
        new_route[idx1], new_route[idx2] = new_route[idx2], new_route[idx1]
    return new_route

# -------------------------------------------------------------
# 3. PARTICLE DEFINITION
# -------------------------------------------------------------
class Particle:
    def __init__(self, customers):
        self.position = copy.copy(customers)
        random.shuffle(self.position)
        self.velocity = []
        self.pbest_position = copy.copy(self.position)
        self.pbest_fitness = evaluate_fitness(self.position)

# -------------------------------------------------------------
# 4. MAIN PSO ROUTING ALGORITHM
# -------------------------------------------------------------
def pso_delivery_router(num_particles=15, iterations=45):
    customers = [1, 2, 3, 4, 5]
    swarm = [Particle(customers) for _ in range(num_particles)]

    gbest_position = copy.copy(swarm[0].position)
    gbest_fitness = swarm[0].pbest_fitness

    for particle in swarm:
        if particle.pbest_fitness < gbest_fitness:
            gbest_fitness = particle.pbest_fitness
            gbest_position = copy.copy(particle.position)

    w = 0.4  # Inertia: Probability of keeping previous swaps
    c1 = 0.9 # Cognitive: Probability of pulling a personal-best swap
    c2 = 0.9 # Social: Probability of pulling a global-best swap

    for it in range(iterations):
        for particle in swarm:
            cognitive_swaps = get_swap_sequence(particle.position, particle.pbest_position)
            social_swaps = get_swap_sequence(particle.position, gbest_position)

            # Combine Swaps into New Velocity based on probabilities
            new_velocity = []
            for swap in particle.velocity:
                if random.random() < w:
                    new_velocity.append(swap)
            for swap in cognitive_swaps:
                if random.random() < c1:
                    new_velocity.append(swap)
            for swap in social_swaps:
                if random.random() < c2:
                    new_velocity.append(swap)

            particle.velocity = new_velocity

            # FIXED: Apply the full updated velocity array to evolve the position
            particle.position = apply_swaps(particle.position, particle.velocity)

            current_fitness = evaluate_fitness(particle.position)

            if current_fitness < particle.pbest_fitness:
                particle.pbest_fitness = current_fitness
                particle.pbest_position = copy.copy(particle.position)

            if current_fitness < gbest_fitness:
                gbest_fitness = current_fitness
                gbest_position = copy.copy(particle.position)

        print(f"Iteration {it+1:02d} -> Best Fleet Route Distance: {gbest_fitness:.2f} km")

    return gbest_position, gbest_fitness

best_route, best_dist = pso_delivery_router()
print("\n=== OPTIMAL SOLUTION FOUND ===")
print(f"Optimal Delivery Sequence: Depot -> {' -> '.join(map(str, best_route))} -> Depot")
print(f"Total Operational Distance: {best_dist:.2f} km")
```
